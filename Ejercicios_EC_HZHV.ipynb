##EJERCICIO 1
Al lanzar dos dados, calcula la probabilidad de que la suma de sus caras sea exactamente 5.

```python
import itertools

# 1. Todos los resultados posibles al lanzar 2 dados (36 en total)
dados = list(itertools.product(range(1, 7), repeat=2))

# 2. Casos favorables que suman 5
favorables = [d for d in dados if sum(d) == 5]

# 3. Probabilidad: favorables / posibles
probabilidad = len(favorables) / len(dados)

print(f"Casos favorables: {favorables}")
print(f"Probabilidad: {probabilidad:.2%} (4/36 = 1/9)")
```

Ejercicio 2

En una contraseña de 3 dígitos, del 0 al 9, calcula la probabilidad de que ninguno de los dígitos se repita.

```python
# 1. Total de combinaciones posibles (del 000 al 999)
total_posibles = 10 * 10 * 10  # 1,000

# 2. Casos favorables sin repetición: 10 * 9 * 8
favorables = 10 * 9 * 8  # 720

# 3. Probabilidad
probabilidad = favorables / total_posibles

print(f"Total de contraseñas posibles: {total_posibles}")
print(f"Contraseñas sin repetir dígito: {favorables}")
print(f"Probabilidad: {probabilidad:.2%} (720/1000 = 18/25)")
```

Ejercicio 3

Tienes 8 canciones distintas y las pones en un orden aleatorio, con todos los órdenes igualmente probables. Calcula la probabilidad de que tu favorita quede primera y la que menos te gusta quede última.

*En el ejemplo de la fila se fijó el primer lugar y el resto se permutó. Aquí quedan fijos el primero y el último. El espacio tiene 8! órdenes.

```python
 import math

# 1. Total de órdenes posibles (8!)
total_posibles = math.factorial(8)  # 40,320

# 2. Casos favorables: 1ª y 8ª fijas, se ordenan las 6 restantes (6!)
favorables = math.factorial(6)      # 720

# 3. Probabilidad
probabilidad = favorables / total_posibles

print(f"Total posibles (8!): {total_posibles}")
print(f"Casos favorables (6!): {favorables}")
print(f"Probabilidad: {probabilidad:.2%} (1/56)")
```

Ejercicio 4

Se lanzan tres dados equilibrados. ¿Qué es más probable, que la suma sea 9 o que la suma sea 10? ¿Por cuántos resultados le gana una suma a la otra?

*Contar “de cuántas maneras se escribe el número”, sin mirar el orden. Por ejemplo 1 + 3 + 5 y 1 + 5 + 3 son resultados distintos. Arma las ternas como en el ejemplo de los dos dados.

```python
import itertools

# 1. Todos los resultados posibles al lanzar 3 dados (216)
dados = list(itertools.product(range(1, 7), repeat=3))

# 2. Casos favorables para cada suma
suman_9 = [d for d in dados if sum(d) == 9]
suman_10 = [d for d in dados if sum(d) == 10]

print(f"Formas de obtener 9:  {len(suman_9)} de 216 ({len(suman_9)/216:.2%})")
print(f"Formas de obtener 10: {len(suman_10)} de 216 ({len(suman_10)/216:.2%})")
print(f"La suma 10 le gana a la suma 9 por {len(suman_10) - len(suman_9)} resultados.")
```

Ejercicio 5

Ana, Bruno, Carla y Diego hacen un amigo secreto: cada quien saca un regalo al azar y todos los repartos son igualmente probables. Calcula la probabilidad de que nadie saque su propio regalo.

```python
import itertools

personas = ["Ana", "Bruno", "Carla", "Diego"]

# 1. Todas las permutaciones de repartir los 4 regalos (24)
repartos = list(itertools.permutations(personas))

# 2. Casos donde la persona en la posición i NO recibe el regalo i
favorables = [
    r for r in repartos
    if all(r[i] != personas[i] for i in range(len(personas)))
]

# 3. Probabilidad
probabilidad = len(favorables) / len(repartos)

print(f"Repartos totales: {len(repartos)}")
print(f"Casos donde nadie se auto-regala: {len(favorables)}")
print(f"Probabilidad: {probabilidad:.2%} (9/24 = 3/8)")
```
