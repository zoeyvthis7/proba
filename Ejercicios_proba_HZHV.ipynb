Se lanza un dado equilibrado. Calcula la probabilidad de obtener un 2, dado que salió un número par.

Ejercicio 2

Se lanzan tres dados. Calcula la probabilidad de que la suma sea 9, dado que las tres caras son distintas.

```python
import itertools

# 1. Todos los resultados posibles de 3 dados (216)
dados = list(itertools.product(range(1, 7), repeat=3))

# 2. Casos posibles: las 3 caras son distintas (120)
distintos = [d for d in dados if len(set(d)) == 3]

# 3. Casos favorables: caras distintas y suma igual a 9 (18)
favorables = [d for d in distintos if sum(d) == 9]

# 4. Probabilidad condicional
probabilidad = len(favorables) / len(distintos)
print(f"Probabilidad: {probabilidad:.2%} ({len(favorables)}/{len(distintos)} = 3/20)")
```

Ejercicio 3

De 100 días:

40 son nublados

De esos, 30 llueve

¿Cuál es la probabilidad de lluvia dado que está nublado?

```python
# Datos
dias_nublados = 40
dias_lluvia_y_nublado = 30

# Probabilidad condicional: P(Lluvia | Nublado)
probabilidad = dias_lluvia_y_nublado / dias_nublados

print(f"Probabilidad: {probabilidad:.2%} ({dias_lluvia_y_nublado}/{dias_nublados} = 3/4)")
```

PROBABILIDAD TOTAL

Ejercicio 1

El 70% de los clientes ven un anuncio y, de ellos, compra el 30%. El 30% no ve el anuncio y, de ellos, compra el 10%. Calcula la probabilidad total de que un cliente compre.

```python
# Datos
p_ve_anuncio = 0.70
p_compra_si_ve = 0.30

p_no_ve_anuncio = 0.30
p_compra_si_no_ve = 0.10

# Teorema de la Probabilidad Total: P(Compra) = P(A)*P(C|A) + P(no A)*P(C|no A)
prob_total_compra = (p_ve_anuncio * p_compra_si_ve) + (p_no_ve_anuncio * p_compra_si_no_ve)

print(f"Probabilidad total de compra: {prob_total_compra:.2%}")
```

Ejercicio 2

Se sabe que:

El 40% de los días son nublados

De esos, el 75% llueve

El 60% de los días no son nublados

De esos, el 10% llueve

¿Cuál es la probabilidad total de que llueva?

```python
# Datos
p_nublado = 0.40
p_llueve_si_nublado = 0.75

p_no_nublado = 0.60
p_llueve_si_no_nublado = 0.10

# Probabilidad total: P(Lluvia) = P(N)*P(LL|N) + P(no N)*P(LL|no N)
prob_total_lluvia = (p_nublado * p_llueve_si_nublado) + (p_no_nublado * p_llueve_si_no_nublado)

print(f"Probabilidad total de que llueva: {prob_total_lluvia:.2%}")
```

Ejercicio 3

Hay dos cajas de audífonos. La primera tiene 2 que sirven y 6 que fallan. La segunda tiene 5 que sirven y 3 que fallan. Se elige una caja al azar y después un audífono. ¿Cuál es la probabilidad de que sirva?

```python
# Probabilidad de elegir cada caja (1/2 cada una)
p_caja1 = 0.5
p_caja2 = 0.5

# Proporción de audífonos que sirven en cada caja
p_sirve_caja1 = 2 / 8  # 2 de 8
p_sirve_caja2 = 5 / 8  # 5 de 8

# Probabilidad total: P(C1)*P(S|C1) + P(C2)*P(S|C2)
prob_total = (p_caja1 * p_sirve_caja1) + (p_caja2 * p_sirve_caja2)

print(f"Probabilidad total de que sirva: {prob_total:.2%} (7/16)")
```

TEOREMA DE BAYES

Ejercicio 1

El 20% de los correos son spam. La palabra «oferta» aparece en el 80% de los correos spam y en el 10% de los correos deseados. Si un correo tiene la palabra «oferta», ¿cuál es la probabilidad de que sea spam?

```python
# Datos
p_spam = 0.20
p_deseado = 0.80

p_oferta_si_spam = 0.80
p_oferta_si_deseado = 0.10

# 1. Probabilidad total de que contenga "oferta"
p_oferta = (p_spam * p_oferta_si_spam) + (p_deseado * p_oferta_si_deseado)

# 2. Teorema de Bayes: P(Spam | Oferta)
p_spam_si_oferta = (p_spam * p_oferta_si_spam) / p_oferta

print(f"Probabilidad de que sea spam: {p_spam_si_oferta:.2%} (2/3)")
```

Ejercicio 2

En una tienda online, el 40% entra por Chrome, el 35% por Safari y el 25% por otros. La tasa de conversión (compra) es del 3% en Chrome, 4% en Safari y 2% en otros.

```python
# Tráfico y conversión
p_chrome, p_safari, p_otros = 0.40, 0.35, 0.25
conv_chrome, conv_safari, conv_otros = 0.03, 0.04, 0.02

# Probabilidad total de compra (0.031)
p_compra = (p_chrome * conv_chrome) + (p_safari * conv_safari) + (p_otros * conv_otros)

# Bayes para cada navegador
probabilidades = {
    "Safari": (p_safari * conv_safari) / p_compra,
    "Chrome": (p_chrome * conv_chrome) / p_compra,
    "Otros":  (p_otros * conv_otros) / p_compra
}

ganador = max(probabilidades, key=probabilidades.get)
print(f"El navegador más probable es {ganador} con {probabilidades[ganador]:.2%}")
```
