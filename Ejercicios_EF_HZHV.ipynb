Ejercicio 1

Simula 10 000 lanzamientos de dos dados y estima la probabilidad de que la suma sea exactamente 5. Compara esa frecuencia con la probabilidad clásica que ya calculaste.



```python
import random

n = 10_000
exitos = 0
for _ in range(n):
    if random.randint(1, 6) + random.randint(1, 6) == 5:
        exitos += 1

print(f"Probabilidad simulada: {exitos / n:.2%}")
print(f"Probabilidad clásica:  {4 / 36:.2%} (4/36 = 1/9)")
```

Ejercicio 2

Simula 10 000 PIN de 4 dígitos, del 0 al 9, y estima la probabilidad de que ninguno se repita. Compara el resultado con la fracción de la parte clásica.

*Usa generar_pin del ejemplo de los tres dígitos iguales, ahora con 4 dígitos.



```python
import random

def generar_pin():
    return [random.randint(0, 9) for _ in range(4)]

n = 10_000
sin_repetir = sum(1 for _ in range(n) if len(set(generar_pin())) == 4)

print(f"Probabilidad simulada: {sin_repetir / n:.2%}")
print(f"Probabilidad clásica:  {(10 * 9 * 8 * 7) / 10_000:.2%} (5040/10000 = 50.40%)")
```

Ejercicio 3

Simula 20 000 lanzamientos de tres dados. Estima la frecuencia de la suma 9 y la de la suma 10. ¿Se nota, en la simulación, cuál de las dos sale más?

*Un ensayo son tres llamadas a lanzar_dado. Lleva dos conteos. La ventaja de una suma sobre la otra es pequeña: si en una corrida casi empatan, sube el número de lanzamientos y vuelve a correr.



```python
import random

def lanzar_dado():
    return random.randint(1, 6)

n = 20_000
c9, c10 = 0, 0

for _ in range(n):
    s = lanzar_dado() + lanzar_dado() + lanzar_dado()
    if s == 9:
        c9 += 1
    elif s == 10:
        c10 += 1

print(f"Suma 9  -> Simulada: {c9 / n:.2%} ({c9} veces) | Clásica: {25/216:.2%}")
print(f"Suma 10 -> Simulada: {c10 / n:.2%} ({c10} veces) | Clásica: {27/216:.2%}")
print(f"¿Se nota cuál sale más?: Sí, la suma 10 le gana a la 9 por {c10 - c9} veces.")
```

Ejercicio 4

Basado en el Ejemplo 3, simula una urna con 3 bolas rojas y 2 azules. Realiza 2,000 extracciones (con reemplazo) y calcula la probabilidad de obtener una azul.



```python
import random

urna = ["roja", "roja", "roja", "azul", "azul"]
n = 2_000
azules = sum(1 for _ in range(n) if random.choice(urna) == "azul")

print(f"Probabilidad simulada de azul: {azules / n:.2%}")
print(f"Probabilidad clásica:          {2 / 5:.2%} (2/5 = 40%)")
```

Ejercicio 5

Una serie son 20 volados. Estima, con 10 000 series, la probabilidad de que aparezcan 5 águilas seguidas en algún momento de la serie.

*Un ensayo no es un volado: son 20 llamadas a lanzar_moneda. Recorre esa serie y cuenta el ensayo si la racha de águilas llega a 5. La estimación es el número de series favorables entre 10 000. Aquí no hay una fracción simple de la parte clásica contra la cual dibujar la línea.





```python
import random

def lanzar_moneda():
    return random.choice(["A", "S"])  # A: Águila, S: Sol

n_series = 10_000
series_con_racha = 0

for _ in range(n_series):
    racha = 0
    hubo_racha = False
    for _ in range(20):
        if lanzar_moneda() == "A":
            racha += 1
            if racha == 5:
                hubo_racha = True
                break
        else:
            racha = 0
    if hubo_racha:
        series_con_racha += 1

print(f"Probabilidad estimada de 5 águilas seguidas: {series_con_racha / n_series:.2%}")
```

Ejercicio 6 — Bonus

Los clientes llegan a una tienda de forma aleatoria (entre 1 y 5 minutos). El cajero tarda entre 2 y 4 minutos en atender. Simula la llegada de 500 clientes y calcula la probabilidad de que un cliente tenga que esperar más de 10 minutos en la fila.

```python
import random

n_clientes = 500
esperaron_mas_10 = 0
tiempo_llegada = 0.0
cajero_libre = 0.0

for _ in range(n_clientes):
    tiempo_llegada += random.uniform(1.0, 5.0)  # Llega entre 1 y 5 min
    inicio_atencion = max(tiempo_llegada, cajero_libre)

    espera = inicio_atencion - tiempo_llegada
    if espera > 10.0:
        esperaron_mas_10 += 1

    servicio = random.uniform(2.0, 4.0)  # Tarda entre 2 y 4 min
    cajero_libre = inicio_atencion + servicio

print(f"Clientes con espera > 10 min: {esperaron_mas_10} de {n_clientes}")
print(f"Probabilidad estimada:        {esperaron_mas_10 / n_clientes:.2%}")
```
